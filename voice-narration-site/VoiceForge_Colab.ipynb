# 🎙️ VoiceForge on Google Colab (free GPU)

Runs the whole VoiceForge site — voice cloning + script narration — on a free Colab GPU.

**Before running:** go to `Runtime → Change runtime type` and pick **T4 GPU** (free tier).

Then run the four cells below in order. The last cell opens the site in a new tab.

Notes:
- The first *Generate narration* click downloads the Chatterbox model weights (~2 GB) — allow a minute or two, once per session.
- Everything runs inside this Colab VM; nothing is sent to any third-party TTS API.
- Colab sessions are temporary: download your narration files before the runtime shuts down.

```
# 2. Install dependencies (~2-3 minutes; Colab already has ffmpeg and CUDA torch)
%pip install -q chatterbox-tts fastapi uvicorn python-multipart
```

```
# 3. Start the VoiceForge server
import subprocess, time, urllib.request

server = subprocess.Popen(["uvicorn", "server.main:app", "--host", "0.0.0.0", "--port", "8080"])
for _ in range(60):
    try:
        urllib.request.urlopen("http://127.0.0.1:8080/api/health")
        print("✔ Server is up")
        break
    except Exception:
        time.sleep(1)
else:
    print("Server did not start — check the output of the previous cells")
```

```
# 4. Open the site (opens a new browser tab served over HTTPS, so mic recording works)
from google.colab import output
output.serve_kernel_port_as_window(8080)
```
